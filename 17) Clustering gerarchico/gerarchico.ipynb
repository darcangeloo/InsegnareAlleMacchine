```python
import numpy as np
from sklearn.datasets import make_blobs
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

np.random.seed(42)
X, y_vero = make_blobs(n_samples=15, centers=3, cluster_std=0.6, random_state=42)
```

```python
Z = linkage(X, method='ward')
```

```python
dendrogram(Z)
plt.xlabel('punti')
plt.ylabel('distanza (Ward)')
plt.show()
```

```python
agg = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = agg.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis')
plt.show()
```

```python
agg_soglia = AgglomerativeClustering(n_clusters=None, distance_threshold=10, linkage='ward')
labels_soglia = agg_soglia.fit_predict(X)
```
