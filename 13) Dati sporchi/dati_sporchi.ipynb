# Rimozione riga

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Feature X":[np.nan, 5, 7, 6],
    'Feature Y': [6, 10, 14, np.nan],
    'Feature Z': [12, 20, np.nan, 4]
})

print(df.dropna())
print(df.drop("Feature Z", axis=1))
```

## Imputazione semplice

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "Feature X":[np.nan, 5, 7, 6],
    'Feature Y': [6, 10, 14, np.nan],
    'Feature Z': [12, 20, np.nan, 4]
})

imputer = SimpleImputer(strategy="mean") # o "median" o "most_frequent" (moda)
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)
print(df_imputed)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Feature X":[np.nan, 5, 7, 6],
    'Feature Y': [6, 10, 14, np.nan],
    'Feature Z': [12, 20, np.nan, 4]
})
print(df.fillna(df.mean()))# ogni colonna riempita con la propria media
print(df.fillna(df.median()))# ogni colonna riempita con la propria mediana
print(df.fillna(df.mode().iloc[0]))  # vedi sotto
```

## Imputazione Avanzata

```python
import pandas as pd
import numpy as np


np.random.seed(42)
n = 10
X = np.round(np.random.uniform(2, 10, n), 1)
Y = np.round(2*X + 3 + np.random.normal(0, 1.5, n), 1)
Z = np.round(X + Y + np.random.normal(0, 1, n), 1)

df = pd.DataFrame({'Feature X': X, 'Feature Y': Y, 'Feature Z': Z})
df.loc[2, 'Feature X'] = np.nan
df.loc[5, 'Feature Y'] = np.nan
df.loc[8, 'Feature Z'] = np.nan
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

reg_imputer = IterativeImputer(random_state=42)
df_reg = pd.DataFrame(reg_imputer.fit_transform(df), columns=df.columns)
print(df_reg)
```

```python
from sklearn.impute import KNNImputer

knn_imputer = KNNImputer(n_neighbors=3)
df_knn = pd.DataFrame(knn_imputer.fit_transform(df), columns=df.columns)
print(df_knn)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer
scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)

knn_imputer = KNNImputer(n_neighbors=3)
df_scaled_imputed = pd.DataFrame(knn_imputer.fit_transform(df_scaled), columns=df.columns)

df_knn_scaled = pd.DataFrame(scaler.inverse_transform(df_scaled_imputed), columns=df.columns)
print(df_knn_scaled)
```
